```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the dataset
mushroom_data = pd.read_csv("mushroom.csv")
```

```python
# Filter out null values in the 'cap-diameter' column
filtered_data = mushroom_data.dropna(subset=['cap-diameter'])
```

```python
# Convert 'cap-diameter' column to float
filtered_data['cap-diameter'] = filtered_data['cap-diameter'].astype(float)
```

```python
# Plot the histogram
plt.figure(figsize=(10, 6))
plt.hist(filtered_data['cap-diameter'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Mushroom Cap Diameters')
plt.xlabel('Cap Diameter (cm)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
# Function to plot bar chart for categorical variables
def plot_bar_chart(column_name, title):
    plt.figure(figsize=(8, 6))
    mushroom_data[column_name].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title(title)
    plt.xlabel(column_name)
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.grid(True)
    plt.show()
```

```python
# Plot bar chart for 'class'
plot_bar_chart('class', 'Distribution of Mushroom Classes')
```

```python
# Plot bar chart for 'cap-shape'
plot_bar_chart('cap-shape', 'Distribution of Mushroom Cap Shapes')
```

```python
# Plot bar chart for 'cap-surface'
plot_bar_chart('cap-surface', 'Distribution of Mushroom Cap Surfaces')
```

```python
# Plot bar chart for 'cap-color'
plot_bar_chart('cap-color', 'Distribution of Mushroom Cap Colors')
```

```python
# Plot bar chart for 'gill-attachment'
plot_bar_chart('gill-attachment', 'Distribution of Gill Attachments')
```

```python
# Plot bar chart for 'gill-spacing'
plot_bar_chart('gill-spacing', 'Distribution of Gill Spacing')
```

```python
# Plot bar chart for 'gill-color'
plot_bar_chart('gill-color', 'Distribution of Gill Colors')
```
