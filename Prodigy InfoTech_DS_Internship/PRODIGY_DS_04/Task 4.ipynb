```python
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob
```

```python
# Load the training data
twitter_training_data = pd.read_csv("twitter_training.csv")
```

```python
# Perform sentiment analysis on the tweet content
twitter_training_data['Sentiment'] = twitter_training_data.iloc[:, -1].apply(lambda x: TextBlob(str(x)).sentiment.polarity)
```

```python
# Count positive, negative, and neutral sentiments
positive_count = (twitter_training_data['Sentiment'] > 0).sum()
negative_count = (twitter_training_data['Sentiment'] < 0).sum()
neutral_count = (twitter_training_data['Sentiment'] == 0).sum()
```

```python
# Plot sentiment distribution
plt.figure(figsize=(8, 6))
plt.hist(twitter_training_data['Sentiment'], bins=10, color='skyblue', edgecolor='black')
plt.xlabel('Sentiment Polarity')
plt.ylabel('Frequency')
plt.title('Sentiment Distribution of Borderlands Tweets')
plt.show()
```

```python
# Pie chart for sentiment distribution
labels = ['Positive', 'Negative', 'Neutral']
sizes = [positive_count, negative_count, neutral_count]
colors = ['lightgreen', 'lightcoral', 'lightskyblue']
explode = (0.1, 0, 0)  # explode 1st slice (positive)
plt.figure(figsize=(8, 6))
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True, startangle=140)
plt.title('Sentiment Distribution of Borderlands Tweets')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
# Display average sentiment score
average_sentiment = twitter_training_data['Sentiment'].mean()
print("Average Sentiment Score:", average_sentiment)
```
